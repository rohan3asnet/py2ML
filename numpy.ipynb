expressing conditional logic as array opterations

```python
x=np.array([1,2,3,4,5])
y=np.array([2,4,6,8,10])
condition=np.array([True,False,True,True,False])
```

now i want to take value from x if condition is true otherwise take value from y

```python
result=[(x if c else y)
        for x,y,c in zip(x,y,condition)]
```

```python
result
```

above code has multiple problem, first it will not be very fast for large arrays, second it will not work with multidimensional arrays.


```python
#numpy.where() can be used to solve the above issue
result=np.where(condition,x,y)
```

```python
result
```


### when to use numpy.where()?
when you want to produce new arrays of value based on another array

```python
h=rng.standard_normal((3,3))
```

```python
h
```

```python
h>0
```

```python
np.where(h>0,6,7)
```

```python
np.where(h>0,6,h)
```
