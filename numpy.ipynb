Basic indexing and slicing


one dimensional array are simple, on the surface they act as python list

```python
arr1
```

```python
arr1[3]
```

```python
arr1[0:3]
```

```python
arr1[2:4]=8
```

```python
arr1
```

array slices are viewed on the original array, data will not be copied, it directly affects the original array

```python
arr1_slice=arr1[2:4]
arr1_slice
```

when i change the value n arr1_slice the value will also change in the orginal array

```python
arr1_slice[0]=9# here i have change the 8 to 9
arr1_slice
arr1
```

```python
arr1_slice[:]=67
arr1
```

in higher dimension, elements at each index are no longer scalars but rather one dimensional arrays

```python
arr2d=([[1,2,3],[2,2,2],[4,5,6]])
arr2d[2]
```

```python
#individual element can be accessed as
arr2d[0][2]
```

in multidimensional array, if you omit later indices,

```python
a2[0][1]
```

```python
a2[0][1][2]
```

```python
a2[1]=33
```
