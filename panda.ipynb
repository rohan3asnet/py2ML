```python
# think series as fixed length ordered dictionary
# because it maps index values to data values
```

```python
# we can convert dictionary to series
wards={'Barahkshetra':5, 'Chandragiri':14, 'Ramdhuni':8, 'Sundarharaichaa':9}
obj4=pd.Series(wards)
```

```python
obj4
```

```python
#series can be converted back to dictionary
obj4.to_dict()
```

```python
#dictionary maa jasari insert gareko hunxam, data tesari nai display hunxa
# tara haami haamlai jun order maa herna man xa tesari display garna sakxam
# by passing index in the order we want
wardss=['Barahkshetra', 'Ramdhuni','Kawaswoti']
obj4=pd.Series(wards, index=wardss)
```

```python
obj4
```

```python
# isna and notna are used to detect missing data
pd.isna(obj4)
```

```python
pd.notna(obj4)
```

```python
obj4.name='Ward Name'
obj4.index.name="ward"
obj4
```
