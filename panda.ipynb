```python
# think series as fixed length ordered dictionary
# because it maps index values to data values
```

```python
# we can convert dictionary to series
wards={'Barahkshetra':5, 'Chandragiri':14, 'Ramdhuni':8, 'Sundarharaichaa':9}
obj4=pd.Series(wards)
```

```python
obj4
```

```python
#series can be converted back to dictionary
obj4.to_dict()
```

```python
#dictionary maa jasari insert gareko hunxam, data tesari nai display hunxa
# tara haami haamlai jun order maa herna man xa tesari display garna sakxam
# by passing index in the order we want
wardss=['Barahkshetra', 'Ramdhuni','Kawaswoti']
obj4=pd.Series(wards, index=wardss)
```

```python
obj4
```

```python
# isna and notna are used to detect missing data
pd.isna(obj4)
```

```python
pd.notna(obj4)
```

```python
obj4.name='Ward Name'
obj4.index.name="ward"
obj4
```

```python
# Dataframe ---> rectangular table of data
# contains ordered, named collection of columns
```

```python
# we can construct our own dataframe
# using dictionary of equal  length list or arrays 

data={"name": ['Ram','Sita', 'Ravana', 'Laxman'],
      "age": [33,31,44,33],
      "height":[6.1,5.7,6.4,6]}

frame=pd.DataFrame(data)
```

```python
frame
```

```python
frame.head()
```

```python
frame.tail()
```

```python
#you can specify the order of columns
pd.DataFrame(frame,columns=['name','height','age'])
```

```python
#passing missing columns result to:
frame2=pd.DataFrame(frame,columns=['name','height','age','kills'])
frame2
```

```python
# following are the ways to retrive columns
```

```python
frame2.age
```

```python
frame2.name
```

```python
frame2['height']
```

```python
# column values can be modified by assignment

#frame2['kills]=[2000,500,4000,1800] 0r 
kills={
    'Ram':2000,
    'Ravana':4000,
    'Sita':500,
    'Laxman':1800
}
frame2["kills"]=frame2['name'].map(kills)
```

```python
frame2
```

```python
# loc and iloc
frame2.loc[1]
```

```python
frame2.iloc[1]
```

```python
value=pd.Series([5,5], index=[0,3])
frame2['bow']=value
frame2
```

```python
# del can delete the column
del frame2['bow']
```
